## Test retrieval

```python
index = client.get_index(
    endpoint_name=VECTOR_SEARCH_ENDPOINT,
    index_name=VECTOR_INDEX_NAME
)

results = index.similarity_search(
    query_text="How do I fix pipeline authentication failure?",
    columns=["chunk"],
    num_results=3
)

print(results)
```
